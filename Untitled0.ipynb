```python
import torch
import torchaudio
import numpy as np
from datasets import load_dataset
from audiomentations import Compose, AddGaussianSNR

# 1. 평가용 Clean 데이터 로드
print("LibriSpeech 데이터 로드 중...")
dataset = load_dataset("openslr/librispeech_asr", "clean", split="validation[:50]")

# 2. 노이즈 합성 파이프라인 구축
degrade_pipeline = Compose([
    AddGaussianSNR(min_snr_db=-5.0, max_snr_db=0.0, p=1.0)
])

# 매핑 함수 수정: .get() 대신 안전한 키 검사 방식으로 path를 가져옵니다.
def apply_tactical_degradation(batch):
    audio_data = batch["audio"]
    audio_array = audio_data["array"]
    sample_rate = audio_data["sampling_rate"]
    audio_path = audio_data["path"] if "path" in audio_data else ""

    # [Step 1] 주파수 대역폭 극단적 제한 (구형 무전기 시뮬레이션: 300Hz ~ 2500Hz)
    tensor_audio = torch.tensor(audio_array).unsqueeze(0).to(torch.float32)
    narrowband_audio = torchaudio.functional.bandpass_biquad(
        tensor_audio, sample_rate=sample_rate, central_freq=1400.0, Q=0.707
    )

    # [Step 2] 노이즈 믹스업 적용
    noisy_audio = degrade_pipeline(samples=narrowband_audio.squeeze().numpy(), sample_rate=sample_rate)

    # 새로운 딕셔너리로 덮어쓰기
    batch["audio"] = {
        "array": noisy_audio,
        "sampling_rate": sample_rate,
        "path": audio_path
    }

    return batch

# 3. 전체 데이터셋에 열화 파이프라인 매핑
print("전장 환경 소음 합성 및 무전 필터 적용 중...")
tactical_dataset = dataset.map(apply_tactical_degradation)

print("전장 환경 데이터셋(Evaluation용) 구축 완료!")
```

```python
import torch
import evaluate
from transformers import WhisperProcessor, WhisperForConditionalGeneration
from tqdm.auto import tqdm

# 1. 평가 지표 (WER) 로드
print("평가 지표(WER) 로드 중...")
wer_metric = evaluate.load("wer")

# 2. Whisper 모델 및 프로세서 로드 (Zero-shot 베이스라인)
print("Whisper-small 모델 로드 중...")
device = "cuda:0" if torch.cuda.is_available() else "cpu"
model_id = "openai/whisper-small"

processor = WhisperProcessor.from_pretrained(model_id)
model = WhisperForConditionalGeneration.from_pretrained(model_id).to(device)

# 3. 추론 및 평가 진행
print("전장 환경 데이터셋 추론 시작...")
predictions = []
references = []

# 앞서 생성한 tactical_dataset (50개 샘플) 순회
for i in tqdm(range(len(tactical_dataset))):
    sample = tactical_dataset[i]
    audio = sample["audio"]
    ground_truth = sample["text"] # 원본 정답 텍스트

    # 모델 입력 형태에 맞게 오디오 전처리
    inputs = processor(audio["array"], sampling_rate=audio["sampling_rate"], return_tensors="pt")
    input_features = inputs.input_features.to(device)

    # 음성 인식(추론) 수행
    with torch.no_grad():
        predicted_ids = model.generate(input_features)

    # 출력된 토큰을 텍스트로 디코딩
    transcription = processor.batch_decode(predicted_ids, skip_special_tokens=True)[0]

    predictions.append(transcription.strip().upper())
    references.append(ground_truth.strip().upper())

# 4. 최종 성능(WER) 계산
wer_score = wer_metric.compute(predictions=predictions, references=references)
print(f"========================================")
print(f"🔥 극한 전장 환경 Zero-shot WER: {wer_score * 100:.2f}%")
print(f"========================================")

# (선택) 첫 번째 샘플의 예측 결과 비교 출력
print(f"[정답]: {references[0]}")
print(f"[예측]: {predictions[0]}")
```

```python
import torch
from dataclasses import dataclass
from typing import Any, Dict, List, Union
from transformers import Seq2SeqTrainer, Seq2SeqTrainingArguments
from peft import LoraConfig, get_peft_model

print("1. 학습을 위한 데이터 전처리 중...")
# 오디오 배열을 Mel-spectrogram으로, 정답 텍스트를 Token ID로 변환
def prepare_dataset(batch):
    audio = batch["audio"]
    batch["input_features"] = processor.feature_extractor(audio["array"], sampling_rate=audio["sampling_rate"]).input_features[0]
    batch["labels"] = processor.tokenizer(batch["text"]).input_ids
    return batch

prepared_dataset = tactical_dataset.map(prepare_dataset, remove_columns=tactical_dataset.column_names)

# Whisper는 입력(오디오)과 출력(텍스트)의 길이가 달라 특수한 데이터 콜레이터(패딩 처리기)가 필요합니다.
@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any
    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        input_features = [{"input_features": feature["input_features"]} for feature in features]
        batch = self.processor.feature_extractor.pad(input_features, return_tensors="pt")
        label_features = [{"input_ids": feature["labels"]} for feature in features]
        labels_batch = self.processor.tokenizer.pad(label_features, return_tensors="pt")
        # 패딩된 부분은 손실 계산에서 제외(-100)
        labels = labels_batch["input_ids"].masked_fill(labels_batch.attention_mask.ne(1), -100)
        if (labels[:, 0] == self.processor.tokenizer.bos_token_id).all().cpu().item():
            labels = labels[:, 1:]
        batch["labels"] = labels
        return batch

data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor=processor)

print("2. LoRA (파라미터 효율적 미세 조정) 설정 중...")
# Whisper 모델의 Attention 블록(q_proj, v_proj)만 타겟으로 삼아 가중치 튜닝
config = LoraConfig(r=8, lora_alpha=32, target_modules=["q_proj", "v_proj"], lora_dropout=0.05, bias="none")
peft_model = get_peft_model(model, config)
peft_model.print_trainable_parameters() # 전체 파라미터의 약 1% 미만만 학습됨을 확인

print("3. 모델 학습 (Fine-Tuning) 시작...")
# 빠른 테스트를 위해 동일한 50개 데이터로 학습 진행 (과적합 유도하여 성능 향상 증명)
training_args = Seq2SeqTrainingArguments(
    output_dir="./tactical_whisper_baseline",
    per_device_train_batch_size=4,
    learning_rate=1e-3,
    num_train_epochs=2,     # 2 에포크만 가볍게 진행
    logging_steps=5,
    remove_unused_columns=False,
    label_names=["labels"],
)

trainer = Seq2SeqTrainer(
    args=training_args,
    model=peft_model,
    train_dataset=prepared_dataset,
    data_collator=data_collator,
)

trainer.train()

trainer.train()

print("4. 학습 완료 모델로 재추론 및 점수 비교...")
peft_model.eval()
ft_predictions = []
ft_references = []

for i in range(len(tactical_dataset)):
    sample = tactical_dataset[i]
    audio = sample["audio"]
    ground_truth = sample["text"]

    inputs = processor(audio["array"], sampling_rate=audio["sampling_rate"], return_tensors="pt")
    input_features = inputs.input_features.to(device)

    with torch.no_grad():
        predicted_ids = peft_model.generate(input_features)

    transcription = processor.batch_decode(predicted_ids, skip_special_tokens=True)[0]
    ft_predictions.append(transcription.strip().upper())
    ft_references.append(ground_truth.strip().upper())

ft_wer_score = wer_metric.compute(predictions=ft_predictions, references=ft_references)
print(f"========================================")
print(f"✅ 파인튜닝 후 베이스라인 WER: {ft_wer_score * 100:.2f}%")
print(f"========================================")
```
